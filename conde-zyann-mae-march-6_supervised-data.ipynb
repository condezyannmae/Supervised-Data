```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
iris = fetch_ucirepo(id=53)

# data (as pandas dataframes)
X = iris.data.features
y = iris.data.targets

# metadata
print(iris.metadata)

# variable information
print(iris.variables)
```

```python
import pandas as pd
import random
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()

X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
rf_model = RandomForestClassifier(n_estimators=100,random_state=42)
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Model Accuracy:", accuracy)
```

```python
results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})
```

```python
results["Actual"] = results["Actual"].apply(lambda x: iris.target_names[x])
results["Predicted"] = results["Predicted"].apply(lambda x: iris.target_names[x])
```

```python
print("\nFlower Predictions:")
print(results.head(10))
```

```python
random_index = random.randint(0, len(X_test)-1)
sample = X_test.iloc[random_index].values.reshape(1,-1)

prediction = rf_model.predict(sample)

print("Random Sample Features:", sample)
print("Predicted Flower:", iris.target_names[prediction][0])
print("Actual Flower:", iris.target_names[y_test.iloc[random_index]])
```
